## 1. Types de base et conversions

Python est **type dynamique** : pas besoin de declarer le type. `type(x)` le donne.

```python
entier = 42
flottant = 3.14
texte = "bonjour"
booleen = True
rien = None

for v in (entier, flottant, texte, booleen, rien):
    print(f"{repr(v):10} -> {type(v).__name__}")
```

```python
# Conversions (casting)
print(int("17") + 3)        # str -> int
print(float("2.5") * 2)     # str -> float
print(str(99) + " ballons") # int -> str
print(list("abc"))          # str -> list
print(bool(0), bool(""), bool([]), bool(1))  # valeurs "falsy"
```

## 2. Chaines de caracteres

Immuables. Le *slicing* `s[debut:fin:pas]` est omnipresent en algo.

```python
s = "algorithme"
print(s[0], s[-1])          # premier, dernier
print(s[0:4])               # tranche "algo"
print(s[::-1])              # inversee
print(s.upper(), s.count("o"))
print("-".join(["a", "b", "c"]))   # jointure
print("a,b,c".split(","))          # decoupe
print(f"longueur = {len(s)}")      # f-string
```

## 3. Structures de donnees

| Structure | Ordonnee | Modifiable | Doublons | Acces |
|---|---|---|---|---|
| `list`  | oui | oui | oui | index |
| `tuple` | oui | non | oui | index |
| `dict`  | oui* | oui | cles uniques | cle |
| `set`   | non | oui | non | -- |

*insertion order depuis Python 3.7*

```python
# Liste : ordonnee, modifiable
nums = [3, 1, 4, 1, 5]
nums.append(9)
nums.sort()
print(nums, "| max =", max(nums), "| somme =", sum(nums))
```

```python
# Dict : cle -> valeur (structure reine en algo)
freq = {}
for c in "banane":
    freq[c] = freq.get(c, 0) + 1
print(freq)
print("cles :", list(freq.keys()))
print("items :", list(freq.items()))
```

```python
# Set : test d'appartenance en O(1), suppression des doublons
vus = set()
doublons = []
for x in [1, 2, 2, 3, 3, 3]:
    if x in vus:
        doublons.append(x)
    vus.add(x)
print("uniques :", vus, "| doublons :", doublons)
```

```python
# Tuple : immuable, sert de cle de dict ou de retour multiple
point = (2, 5)
x, y = point            # unpacking
print(x, y)
def min_max(lst): return min(lst), max(lst)
print(min_max([4, 9, 1]))
```

## 4. Conditions et boucles

`enumerate` (index + valeur), `zip` (parcours parallele) et `range` sont
indispensables en algo.

```python
for i, val in enumerate(["a", "b", "c"]):
    print(i, val)

for nom, age in zip(["Ana", "Bob"], [30, 25]):
    print(nom, age)

# while avec else (else s'execute si pas de break)
n = 7
for d in range(2, n):
    if n % d == 0:
        print(f"{n} divisible par {d}")
        break
else:
    print(f"{n} est premier")
```

## 5. Fonctions

Arguments positionnels, par defaut, `*args`, `**kwargs`, et `lambda`.

```python
def salue(nom, message="Bonjour"):
    return f"{message}, {nom}!"

print(salue("Ana"))
print(salue("Bob", "Salut"))

# *args / **kwargs
def total(*nombres, **options):
    s = sum(nombres)
    return s * options.get("facteur", 1)

print(total(1, 2, 3))
print(total(1, 2, 3, facteur=10))

# lambda : fonction anonyme (tri par cle)
paires = [("a", 3), ("b", 1), ("c", 2)]
print(sorted(paires, key=lambda p: p[1]))
```

## 6. Comprehensions

La facon *pythonique* de construire listes, dicts et sets en une ligne.

```python
carres = [x**2 for x in range(6)]
pairs = [x for x in range(10) if x % 2 == 0]
print(carres)
print(pairs)

# dict comprehension
mots = ["chat", "chien", "oiseau"]
longueurs = {m: len(m) for m in mots}
print(longueurs)

# set comprehension
print({x % 3 for x in range(10)})
```

## 7. Classes et objets

`__init__` (constructeur), `self`, methodes. Utile pour `ListNode`,
`TreeNode`, `UnionFind` dans les algos.

```python
class Pile:
    def __init__(self):
        self._items = []
    def push(self, x):
        self._items.append(x)
    def pop(self):
        return self._items.pop()
    def est_vide(self):
        return len(self._items) == 0
    def __repr__(self):
        return f"Pile({self._items})"

p = Pile()
p.push(1); p.push(2); p.push(3)
print(p, "| pop ->", p.pop(), "|", p)
```

## 8. Exceptions

`try / except / else / finally`. Gerer les erreurs sans planter.

```python
def diviser(a, b):
    try:
        resultat = a / b
    except ZeroDivisionError:
        return "division par zero !"
    except TypeError as e:
        return f"type invalide : {e}"
    else:
        return resultat        # si pas d'erreur
    finally:
        pass                   # toujours execute

print(diviser(10, 2))
print(diviser(10, 0))
print(diviser(10, "x"))
```

## 9. Iterateurs et generateurs

Un **generateur** (`yield`) produit les valeurs a la demande, sans tout
stocker en memoire -- precieux sur de grandes sequences.

```python
def compte_a_rebours(n):
    while n > 0:
        yield n
        n -= 1

for x in compte_a_rebours(3):
    print(x)

# expression generatrice (parentheses au lieu de crochets)
somme_carres = sum(x**2 for x in range(1000))
print("somme des carres :", somme_carres)
```

## 10. Idiomes utiles pour les algos

Le "boite a outils" a avoir dans les doigts.

```python
from collections import Counter, defaultdict, deque

# Counter : frequences en une ligne
print(Counter("mississippi"))

# defaultdict : evite les KeyError
graphe = defaultdict(list)
graphe[0].append(1)
graphe[0].append(2)
print(dict(graphe))

# deque : file/pile efficace aux deux bouts (BFS !)
q = deque([1, 2, 3])
q.appendleft(0)
print(q, "| popleft ->", q.popleft())
```

```python
# Autres idiomes frequents
nums = [5, 2, 8, 1]

print(sorted(nums, reverse=True))     # tri decroissant
print(min(nums), max(nums))
print(any(x > 7 for x in nums))       # au moins un
print(all(x > 0 for x in nums))       # tous
a, b = 1, 2
a, b = b, a                           # echange sans variable temporaire
print(a, b)
print([0] * 5)                        # tableau initialise
grille = [[0] * 3 for _ in range(2)]  # matrice 2x3 (attention aux copies !)
print(grille)
```

## 11. Pieges frequents

Les erreurs classiques qui coutent cher en entretien.

```python
# PIEGE 1 : argument par defaut mutable (partage entre appels !)
def mauvais(x, liste=[]):     # NE PAS FAIRE
    liste.append(x)
    return liste
print(mauvais(1), mauvais(2))  # [1] puis [1, 2] : surprise !

def bon(x, liste=None):        # bonne pratique
    if liste is None:
        liste = []
    liste.append(x)
    return liste
print(bon(1), bon(2))          # [1] puis [2]
```

```python
# PIEGE 2 : copie de matrice avec * (references partagees)
mauvaise = [[0] * 3] * 2       # les 2 lignes sont le MEME objet
mauvaise[0][0] = 9
print("mauvaise :", mauvaise)  # [[9,0,0],[9,0,0]] !!

bonne = [[0] * 3 for _ in range(2)]
bonne[0][0] = 9
print("bonne    :", bonne)     # [[9,0,0],[0,0,0]]
```

```python
# PIEGE 3 : modifier une liste pendant qu'on l'itere
nums = [1, 2, 3, 4, 5]
# Faire une copie pour iterer, ou construire une nouvelle liste
nums = [x for x in nums if x % 2 == 0]
print(nums)

# PIEGE 4 : is vs == (identite vs egalite)
a = [1, 2]; b = [1, 2]
print(a == b)   # True  : memes valeurs
print(a is b)   # False : objets differents
```
