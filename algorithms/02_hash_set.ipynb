# 2. Hash Set
Ensemble de valeurs uniques avec test d'appartenance O(1) en moyenne.

Cas d'usage typiques :
- Detecter les doublons
- Trouver l'intersection / union
- Parcours avec memoire des elements deja vus

```python
def contains_duplicate(nums: list[int]) -> bool:
    seen: set[int] = set()
    for num in nums:
        if num in seen:
            return True
        seen.add(num)
    return False
```

```python
def first_unique_char(s: str) -> int:
    seen: set[str] = set()
    repeated: set[str] = set()

    for i, ch in enumerate(s):
        if ch in seen:
            repeated.add(ch)
        else:
            seen.add(ch)

    for i, ch in enumerate(s):
        if ch not in repeated:
            return i
    return -1
```

```python
def intersection(nums1: list[int], nums2: list[int]) -> list[int]:
    set1 = set(nums1)
    return list({num for num in nums2 if num in set1})
```

```python
def longest_consecutive(nums: list[int]) -> int:
    num_set = set(nums)
    best = 0

    for num in num_set:
        if num - 1 in num_set:
            continue
        current = num
        length = 1
        while current + 1 in num_set:
            current += 1
            length += 1
        best = max(best, length)
    return best
```

```python
def _sum_squares(n: int) -> int:
    total = 0
    while n:
        n, digit = divmod(n, 10)
        total += digit * digit
    return total
```

```python
def is_happy(n: int) -> bool:
    seen: set[int] = set()
    while n != 1 and n not in seen:
        seen.add(n)
        n = _sum_squares(n)
    return n == 1
```

## Tests

Execute la cellule ci-dessous : elle valide toutes les fonctions avec des assertions. Un message `OK` confirme que tout passe.

```python
def run_tests() -> None:
    assert contains_duplicate([1, 2, 3, 1]) is True
    assert contains_duplicate([1, 2, 3, 4]) is False
    assert first_unique_char("leetcode") == 0
    assert first_unique_char("aabb") == -1
    assert sorted(intersection([1, 2, 2, 1], [2, 2])) == [2]
    assert longest_consecutive([100, 4, 200, 1, 3, 2]) == 4
    assert is_happy(19) is True
    assert is_happy(2) is False
    print("Tous les tests hash_set OK")

run_tests()
```
